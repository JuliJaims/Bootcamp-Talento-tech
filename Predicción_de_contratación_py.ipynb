## Imputaciones


```python
muestra['adjudicado'] = muestra.adjudicado.apply(lambda x: 0 if x == "No" else 1)
```

```python
######## Tabla de comparación
sns.boxplot(y = 'fase', x = 'adjudicado', data = muestra, palette='winter')
```

```python
round(muestra.groupby("fase") ['adjudicado'].mean(), 1)
```

```python
def impute_adj(cols):
    fase = cols[0]
    adjudicado = cols[1]

    if pd.isnull(fase):
        if adjudicado == 1:
            return "Presentación de oferta"
        elif adjudicado == 0:
            return "Fase de Selección (Presentación de ofertas) "
        else:
            return "Fase de Selección (Presentación de ofertas) "
    else:
        return fase
```

```python
######### Aplicamos la función de imputación para realizar cambios
muestra["fase"] = muestra[['fase', 'adjudicado']].apply(impute_adj, axis=1)
muestra.head(2)
```

```python
sns.heatmap(muestra.isnull(), yticklabels  = False, cbar = False, cmap = 'viridis')
```

# **Entrenar el modelo**

```python
# Bibliotecas adicionales
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
```

```python
# Convertir variables categóricas a numéricas con LabelEncoder
label_encoders = {}
for column in X.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    X[column] = le.fit_transform(X[column].astype(str))
    label_encoders[column] = le
```

```python
# Modelo Random Forest
clf = RandomForestClassifier(random_state=42)
clf.fit(X_train, y_train)

# Predicción en el conjunto de prueba
y_pred = clf.predict(X_test)
```
